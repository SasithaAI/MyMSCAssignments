```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
SriLanka_Dengue_df = pd.read_csv('Sri_lanka_dengue_cases_weather_weekly_2007_2024_.csv')
```

```python
# Display Basic Information about the Dataset
print("\n ===Sri_lanka_dengue_cases_Overview=== \n")
print("\nDataset Shape: ", SriLanka_Dengue_df.shape)
print("\nDataset Columns: \n", SriLanka_Dengue_df.columns)
print("\nDataset First Records: \n", SriLanka_Dengue_df.head(5))
```

```python
print("\nDataset Info: \n")
print(SriLanka_Dengue_df.info())
print("\nDataset Description: \n")
print(SriLanka_Dengue_df.describe())
```

```python
print("\n ===Missing Values in the Dataset=== \n")
print(SriLanka_Dengue_df.isnull().sum())
```

```python
# Distribution of Dengue Cases
plt.figure(figsize=(10, 6))
sns.histplot(SriLanka_Dengue_df['Number_of_Cases'], bins=30, kde=True)
plt.title('Distribution of Dengue Cases')
plt.xlabel('Number of Dengue Cases')
plt.ylabel('Frequency')
plt.show()
```

### Categorical Columns Analyasis

```python
#1. Distric vs Number of Cases
plt.figure(figsize=(12, 6))
sns.boxplot(x='District', y='Number_of_Cases', data=SriLanka_Dengue_df)
plt.title('Dengue Cases by District')
plt.xlabel('District')
plt.ylabel('Number of Dengue Cases')
plt.xticks(rotation=45)
plt.show()
```

#### Numaric Columns Correlations

```python
#Numaric Columns Correlations
numaric_cols = SriLanka_Dengue_df.select_dtypes(include=['int64', 'float64']).columns
plt.figure(figsize=(12, 8))
correlation_matrix = SriLanka_Dengue_df[numaric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numaric Features')
plt.show()
```

```python
# Box Plots For Numarical Columns
import math

n_cols = 3
n_rows = math.ceil(len(numaric_cols) / n_cols)

plt.figure(figsize=(15, 5 * n_rows))
for i, col in enumerate(numaric_cols):
    plt.subplot(n_rows, n_cols, i + 1)
    sns.boxplot(x='Number_of_Cases', y=col, data=SriLanka_Dengue_df, palette='Set3')
    plt.title(f'Box Plot of {col} by Number_of_Cases')
    plt.xlabel('Number_of_Cases (0 = No, 1 = Yes)')
plt.tight_layout()
plt.show()
```

#### Correlation of Numarical Columns With Target Column [Number_of_Cases] 

```python
# Correlation of Numarical Columns With Target Column [Number_of_Cases] 
target_col = 'Number_of_Cases'
correlations = SriLanka_Dengue_df[numaric_cols].corrwith(SriLanka_Dengue_df[target_col])
#Drop the target column from the correlations series
correlations = correlations.drop(target_col)
correlation_df = pd.DataFrame(correlations, columns=['Correlation_with_Number_of_Cases'])
correlation_df = correlation_df.sort_values(by='Correlation_with_Number_of_Cases', ascending=False)
print("\nCorrelation of Numarical Columns with Number_of_Cases:\n")
print(correlation_df)
```

```python
plt.figure(figsize=(40, 20))  # bigger figure for readability
correlation_df.plot(kind='bar', color='teal')

plt.title('Correlation of Numeric Features with Number_of_Cases', fontsize=14, pad=15)
plt.ylabel('Correlation Coefficient', fontsize=12)
plt.xticks(rotation=90, fontsize=9)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.show()
```

```python
## Pair Plot for Numarical Features
sns.pairplot(SriLanka_Dengue_df[numaric_cols])
plt.suptitle('Pair Plot of Numarical Features', y=1.02)
plt.show()
```
