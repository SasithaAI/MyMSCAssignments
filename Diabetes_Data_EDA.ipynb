#### Numaric Columns Correlations

```python
# Numerical Column Correlations
numerical_cols = Diabetes_data_df.select_dtypes(include=['int64', 'float64']).columns
correlation_matrix = Diabetes_data_df[numerical_cols].corr()
plt.figure(figsize=(10,8))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', square=True)
plt.title('Correlation Matrix of Numerical Features')
plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90,
    fontsize=8
)
plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    fontsize=8
)

# Add gridlines for clarity
plt.grid(False)
plt.tight_layout()
plt.show()
```

```python
# Box Plots For Numarical Columns
plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols):
    plt.subplot(4, 3, i + 1)
    sns.boxplot(x='diagnosed_diabetes', y=col, data=Diabetes_data_df, palette='Set3')
    plt.title(f'Box Plot of {col} by Diagnosed Diabetes')
    plt.xlabel('Diagnosed Diabetes (0 = No, 1 = Yes)')
    plt.tight_layout()
plt.show()
```

#### Correlation of Numarical Columns With Target Column [Dignostioc Od Dibatics] 

```python
# Correlation Numarical Columns Vs Dignostioc Od Dibatics 
correlation_with_target = correlation_matrix['diagnosed_diabetes'].drop('diagnosed_diabetes')
print("\n ===Correlation of Numerical Columns With Target Column [Diagnosed Diabetes]=== \n")
print(correlation_with_target.sort_values(ascending=False))
```

```python
# Sort The Correlation Values
Dibetes_corr_sorted = correlation_with_target.sort_values(ascending=False)

#Bottom 6 Correlation Columns
print("\n ===Bottom 6 Correlation Columns With Diagnosed Diabetes=== \n")
bottom6_corr_Diagnosed_Diabetes = Dibetes_corr_sorted.tail(6)
print(bottom6_corr_Diagnosed_Diabetes)
```

```python
#Top 6 Correlation Columns
print("\n ===Top 6 Correlation Columns With Diagnosed Diabetes=== \n")
top6_corr_Diagnosed_Diabetes = Dibetes_corr_sorted.head(6)
print(top6_corr_Diagnosed_Diabetes)
```

```python
plt.figure(figsize=(12, 8))  # bigger figure for readability
correlation_with_target.plot(kind='bar', color='teal')

plt.title('Correlation of Numeric Features with Diagnosed Diabetes (yes/no)', fontsize=14, pad=15)
plt.ylabel('Correlation Coefficient', fontsize=12)
plt.xticks(rotation=90, fontsize=9)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.show()
```

```python
top_n = 6
top_features = correlation_with_target.head(top_n).index.tolist() + ['diagnosed_diabetes']

# Create pairplot for top features
sns.pairplot(Diabetes_data_df[top_features], diag_kind='kde', corner=True)
plt.suptitle(f'Pair Plot: Top {top_n} Features vs Diagnosed Diabetes', fontsize=14, y=1.02)
plt.show() 
```
