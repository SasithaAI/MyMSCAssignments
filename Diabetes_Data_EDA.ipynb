#### Catagorical Veriables Correlation 

```python
def cramers_v(x, y):
    """Calculate Cramér's V statistic for categorical-categorical association."""
    confusion_matrix = pd.crosstab(x, y)
    chi2 = stats.chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1)) / (n-1))
    rcorr = r - ((r-1)**2) / (n-1)
    kcorr = k - ((k-1)**2) / (n-1)
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))

# Example: Compute for all categorical variables vs target
target = 'diagnosed_diabetes'
categorical_cols = Diabetes_data_df.select_dtypes(include='object').columns.tolist()

corr_results = {}
for col in categorical_cols:
    corr_results[col] = cramers_v(Diabetes_data_df[col], Diabetes_data_df[target])

# Convert to DataFrame
corr_df = pd.DataFrame(list(corr_results.items()), columns=['Variable', 'CramersV']).sort_values(by='CramersV', ascending=False)

# Set figure size and style
plt.figure(figsize=(10, 6))
bars = plt.barh(
    corr_df['Variable'],
    corr_df['CramersV'],
    color=plt.cm.viridis(corr_df['CramersV'] / corr_df['CramersV'].max())
)

# Add value labels next to each bar
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.01,               # small offset to the right
             bar.get_y() + bar.get_height()/2,
             f"{width:.2f}",
             va='center', fontsize=10, color='black')

# Titles and labels
plt.title("Categorical Variables vs Diagnosed Diabetes (Cramér’s V)", fontsize=14, fontweight='bold')
plt.xlabel("Cramér’s V (Association Strength)", fontsize=12)
plt.ylabel("Categorical Variable", fontsize=12)

# Grid and layout
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.xlim(0, 1)  # since Cramér’s V ranges 0–1
plt.gca().invert_yaxis()  # keep strongest on top
plt.tight_layout()

plt.show()
```

#### Numaric Columns Correlations

```python
# Select only numerical columns
numerical_cols = Diabetes_data_df.select_dtypes(include=['int64', 'float64']).columns

# Compute correlation matrix
correlation_matrix = Diabetes_data_df[numerical_cols].corr()

# Mask the diagonal and upper triangle
mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))

# Plot heatmap without diagonal
plt.figure(figsize=(20,10))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap='coolwarm',
    square=True,
    mask=mask
)

plt.title('Correlation Matrix of Numerical Features (without Diagonal)')
plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90,
    fontsize=8
)
plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    fontsize=8
)

```

## One Hot Encoding the Catagorical Veriables And Check the Correlation With Target[ Diagnosed Diabetes]

```python
# One Hot Encoding the Catagorical Veriables And Check the Correlation With Target[ Diagnosed Diabetes]
print("\n ===One Hot Encoding the Catagorical Veriables And Check the Correlation With Target[ Diagnosed Diabetes]=== \n")       
categorical_cols = Diabetes_data_df.select_dtypes(include='object').columns
Diabetes_data_encoded = pd.get_dummies(Diabetes_data_df, columns=categorical_cols, drop_first=True)
correlation_matrix_encoded = Diabetes_data_encoded.corr()
correlation_with_target_encoded = correlation_matrix_encoded['diagnosed_diabetes'].drop('diagnosed_diabetes')
correlation_with_target_encoded_sorted = correlation_with_target_encoded.sort_values(ascending=False)
print(correlation_with_target_encoded_sorted)
```

```python
mask = np.triu(np.ones_like(correlation_matrix_encoded, dtype=bool))

plt.figure(figsize=(18, 14))  # larger canvas

# Plot heatmap
sns.heatmap(
    correlation_matrix_encoded,
    mask=mask,
    cmap='coolwarm',
    annot=True,
    fmt=".2f",
    square=True,
    linewidths=0.5,
    annot_kws={"size": 8},  # smaller text inside boxes
    cbar_kws={"shrink": 0.8, "label": "Correlation Coefficient"}
)

# Title
plt.title(
    'Correlation Matrix of Encoded Numerical Features (Upper Triangle)',
    fontsize=18,
    fontweight='bold',
    pad=25
)

# X-axis labels
plt.xticks(
    rotation=45,        # angled to prevent overlap
    ha='right',         # align to right edge
    fontsize=9
)

# Y-axis labels
plt.yticks(
    rotation=0,         # keep readable horizontally
    fontsize=9
)

# Adjust spacing around the heatmap
plt.subplots_adjust(left=0.25, bottom=0.25, right=0.95, top=0.9)

plt.tight_layout()
plt.show()
```

```python
# Corelation Posyive Correlations vs  Target Sorted  in a pie Chart
print("\n=== Correlation Target Sorted in a Pie Chart ===\n")

# Sort correlations descending
correlation_values = correlation_with_target_encoded.sort_values(ascending=False)

# Keep only positive correlations
correlation_values = correlation_values[correlation_values > 0.05]


# Use a refined Seaborn color palette (balanced and readable)
colors = sns.color_palette("Set3", len(correlation_values))

# Create a larger figure for better readability
fig, ax = plt.subplots(figsize=(35, 35))
plt.subplots_adjust(left=0.05, right=0.95, top=0.9, bottom=0.05)

# Explode all slices slightly to create spacing between them
explode = [0.07 for _ in correlation_values]

# Draw the pie chart
wedges, texts, autotexts = ax.pie(
    correlation_values,
    labels=None,  # manual labeling below
    autopct='%1.1f%%',
    startangle=140,
    colors=colors,
    pctdistance=0.77,
    explode=explode,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2.2},
    textprops={'fontsize': 17, 'color': 'black'}
)

# Label style and arrow settings
bbox_props = dict(boxstyle="round,pad=0.5", fc="white", ec="gray", lw=0.8, alpha=0.9)
arrowprops = dict(arrowstyle="-", color="gray", lw=1.3, shrinkA=0, shrinkB=5)

# Add labels neatly outside the pie
for i, wedge in enumerate(wedges):
    angle = (wedge.theta2 + wedge.theta1) / 2
    x = np.cos(np.deg2rad(angle))
    y = np.sin(np.deg2rad(angle))
    horizontalalignment = 'left' if x > 0 else 'right'
    connectionstyle = f"angle,angleA=0,angleB={angle}"
    ax.annotate(
        correlation_values.index[i],
        xy=(x, y),
        xytext=(1.6*np.sign(x), 1.5*y),  # more spacing between label and pie
        horizontalalignment=horizontalalignment,
        fontsize=17,
        fontweight='bold',
        bbox=bbox_props,
        arrowprops=dict(arrowstyle="-", color="gray", lw=1.2, connectionstyle=connectionstyle)
    )

# Add a white circle for a clean donut look
centre_circle = plt.Circle((0, 0), 0.63, fc='white')
ax.add_artist(centre_circle)

# Optional: Add a centered text label
ax.text(0, 0, 'Diabetes\nCorrelations',
        ha='center', va='center', fontsize=20, fontweight='bold', color='gray')

# Title
plt.title('Positive Feature Correlations with Diagnosed Diabetes (Yes/No)',
          fontsize=25, fontweight='bold', pad=35)

# Style polish
ax.axis('equal')  # perfect circle
sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```
